## Importing the Libraries

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

- ImageDataGenerator is a class in Keras that helps generate augmented batches of image data during the training of a NN. Data augmentation involves applying random transformations to the input data, such as rotation, zooming, and horizontal/vertical flipping. This helps to artificially increase the diversity of the training dataset, which can improve the generalization of the model

## Data Preprocessing

### Preprocessing the Training Set

- After, runnung this code, the training_set can be used to feed batches of augmented images and their labels to your model during training.

```python
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True,
                                   validation_split=0.2) # Specify 20% for validation

training_set = train_datagen.flow_from_directory('./LeavesIdentifier/Images/Train',
                                                 target_size = (64, 64),
                                                 batch_size = 32,
                                                 class_mode = 'categorical',
                                                 subset='training')  # Specify that this is the training set

validation_set = train_datagen.flow_from_directory('./LeavesIdentifier/Images/Train',
                                                   target_size=(64, 64),
                                                   batch_size=32,
                                                   class_mode='categorical',
                                                   subset='validation')  # Specify that this is the validation set
```

- **`shear_range=0.2`**: Applies a random shear transformation to images, creating a “slanted” effect on objects in the image by shifting them along the x-axis. A shear of `0.2` means up to a 20% shear.

- **`zoom_range=0.2`**: Randomly zooms into images by up to 20%. This helps the model generalize better by making it invariant to slight changes in object scale.

- **`horizontal_flip=True`**: Randomly flips images horizontally. This augmentation helps the model learn to recognize objects regardless of their orientation (e.g., flipping a leaf left-to-right).

- **`validation_split=0.2`**: Reserves 20% of the data for validation, while 80% is used for training. This split allows for better performance evaluation on unseen data without using an external validation dataset.


- **Directory (`./LeavesIdentifier/Images/Train`)**: Specifies the relative path to the training images. This directory is expected to contain subdirectories for each class (e.g., each type of leaf).

- **`target_size=(64, 64)`**: Resizes all images to 64x64 pixels. This standardized input size is required by most deep learning models. Adjusting image size can impact model performance and accuracy, so it is often chosen based on model needs.

- **`batch_size=32`**: The number of images loaded and processed in memory at a time. A batch size of 32 is standard and helps balance memory use and training speed.

- **`class_mode='categorical'`**: Specifies that labels are categorical (i.e., multiple classes), resulting in one-hot encoded labels (e.g., if there are 5 classes, each label will be an array of 5 values with one “hot” 1).

- **`subset='training'`**: Uses only the 80% subset designated for training based on `validation_split=0.2` in `ImageDataGenerator`. 


### Preprocessing the Test Set

- Setting up an image data generator for your test set using the flow_from_directory function from Keras

```python
test_datagen = ImageDataGenerator(rescale = 1./255)
test_set = test_datagen.flow_from_directory('./LeavesIdentifier/Images/Test',
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'categorical')
```

- target_size=(64, 64): It's important to use the same target size as used during training.
- After running this code, the test_set can be used to evaluate your neural network on the test data

## Building the CNN Model

- Create a CNN model for image classification

### Step 1 - Initialising the Model

```python
Model = tf.keras.models.Sequential()
```

- **Sequential Model**: This type of model allows layers to be added one after another in a linear stack, which is ideal for most convolutional neural networks (CNNs) as they usually have a sequential structure.

### Step 2 - Adding First Convolution Layer 
The number of filters directly affects the accuracy level of the model, increasing it will increase the accuracy.

```python
Model.add(tf.keras.layers.Conv2D(filters=16, kernel_size=5, activation='relu', input_shape=[64, 64, 3]))
```

- **Conv2D Layer**: This layer performs convolution operations to detect features (such as edges, shapes, and textures) from input images.
  - **`filters=16`**: Specifies the number of filters (or kernels) to apply, which is 16 here. Each filter detects different features from the input.
  - **`kernel_size=5`**: The dimensions of each filter are 5x5 pixels. Larger kernels capture more spatial information, while smaller ones focus on finer details.
  - **`activation='relu'`**: Applies the ReLU (Rectified Linear Unit) activation function, which helps the model learn nonlinear patterns. It replaces negative values with zero, helping the model avoid the vanishing gradient problem.
  - **`input_shape=[64, 64, 3]`**: The input shape specifies the dimensions of the input images: 64x64 pixels with 3 color channels (RGB). This is only set for the first layer.

### Step 3 - Pooling the First Layer

```python
Model.add(tf.keras.layers.MaxPool2D(pool_size=4, strides=2))
```

- **MaxPool2D Layer**: Reduces the spatial dimensions of the feature maps by selecting the maximum value in each pooling window, which helps reduce the computation and controls overfitting by reducing the model’s sensitivity to small translations.
  - **`pool_size=4`**: Specifies a 4x4 window, meaning the pooling operation will condense a 4x4 region to a single maximum value.
  - **`strides=2`**: The window moves 2 pixels at a time, which reduces the size of the feature map while retaining important features. 

### Step 4 - Adding a Second Convolutional Layer

- Add more layers for deeper understanding

```python
Model.add(tf.keras.layers.Conv2D(filters=16, kernel_size=5, activation='relu'))
```

- you can increase the number of filters for deeper feature extraction.Further increasing the complexity by adding more filters.

### Step 5 - Pooling the Second Layer

```python
Model.add(tf.keras.layers.MaxPool2D(pool_size=4, strides=2))
```

### Step 6 - Flattening

```python
Model.add(tf.keras.layers.Flatten())
```

- **Flatten Layer**: Converts the 2D feature maps into a 1D vector (flattening) to prepare it for the dense (fully connected) layers. This transformation allows the dense layers to process the data as a single vector of features.

### Step 7 - Full Connection

- This adds a fully connected (dense) layer to the model.Dense layers for profound connections between leaf features

```python
Model.add(tf.keras.layers.Dense(units=32, activation='relu'))
```

- **Dense Layer**: A fully connected layer where each neuron is connected to every neuron in the previous layer.
  - **`units=32`**: Specifies 32 neurons in this layer, which enables the network to learn 32 unique, high-level features.
  - **`activation='relu'`**: Uses ReLU to introduce non-linearity, allowing the model to learn complex relationships.

### Step 8 - Output Layer
For classes greater than two, a softmax activation function is used

```python
Model.add(tf.keras.layers.Dense(units=6, activation='softmax')) # units = number of classes
```

- **Dense Layer**: This is the output layer of the model.
  - **`units=6`**: Sets the number of neurons to 6, which should match the number of classes in the classification task. Each neuron represents the probability of the input image belonging to one of the classes.
  - **`activation='softmax'`**: Uses the softmax activation function, which produces a probability distribution over the 6 classes by ensuring the output values sum to 1. This is ideal for multi-class classification tasks.

### Step 9 - Compiling the CNN
In case of N-classes, categorical crossentropy method is used, with nadam optimizer

```python
Model.compile(loss='categorical_crossentropy', optimizer='nadam',metrics=['accuracy'])
```

- Compile your model with optimization and loss calculation
- **`loss='categorical_crossentropy'`**: The categorical cross-entropy loss function is used for multi-class classification tasks with one-hot encoded labels. It measures the difference between the true label distribution and the predicted distribution.
- **`optimizer='nadam'`**: Nadam (Nesterov-accelerated Adaptive Moment Estimation) is an optimization algorithm that combines momentum and adaptive learning rates. It helps with faster convergence and smoother training, especially for deep networks.
- **`metrics=['accuracy']`**: Tracks accuracy as the primary evaluation metric. It measures the fraction of correctly classified samples during training and validation.

## Training the CNN and Evaluation

- The epochs parameter specifies the number of times the model will iterate over the entire training dataset. 

```python
history=Model.fit(x = training_set, validation_data = validation_set, epochs = 35)
```

**Convolutional Layers**
- **Purpose**: The primary goal of convolutional layers is to learn and detect features within the image, such as edges, textures, shapes, and complex patterns.
- **Optimization**: The convolutional layers learn by optimizing the values in their filters (or kernels). 
  - Each filter in a convolutional layer starts with random values, but during training, these values (also called kernel weights) are optimized to detect specific patterns that are useful for distinguishing among the classes.
  - As the network trains, the optimization process (backpropagation) adjusts these kernel weights based on how well they help the network predict the correct class.
  - For example, earlier convolutional layers might learn simple features like edges or colors, while later layers capture more complex patterns, such as parts of shapes or even entire object components (like leaves or textures in your case).

**Dense Layers (Fully Connected Layers)**
- **Purpose**: The dense (fully connected) layers take the high-level features extracted by the convolutional layers and learn the relationships between these features to perform the final classification.
- **Optimization**: In these layers, the network optimizes the weights connecting each neuron in one layer to every neuron in the next layer.
  - These weights are adjusted to learn how combinations of high-level features (detected by the convolutional layers) correlate with each output class.
  - The final dense layer, with the softmax activation, provides the classification probabilities for each class based on the learned features.

**How This Works Together**
- The **convolutional layers** act as feature detectors, learning to identify patterns relevant to the classification task by adjusting the filter weights.
- The **dense layers** act as the decision-making layers, learning to combine these patterns effectively to classify the image correctly.
 


## Making a Prediction

```python
import numpy as np
import os  # It is used for file and path manipulation.
from keras.preprocessing import image

path = r'./LeavesIdentifier/Unkown/Leaf'
for img in os.listdir(path):
    # Load and preprocess the image
    test_image = image.load_img(path + "/" + img, target_size=(64, 64))
    test_image = image.img_to_array(test_image)
    test_image = np.expand_dims(test_image, axis=0)

    # Make the prediction
    predict = Model.predict(test_image)
    
    # Use np.argmax to find the index with the highest probability
    predicted_class = np.argmax(predict[0])

    # Map the index to the class label
    if predicted_class == 0:
        prediction = 'Cherry Blossom'
    elif predicted_class == 1:
        prediction = 'Gingko'
    elif predicted_class == 2:
        prediction = 'Japanese Maple'
    elif predicted_class == 3:
        prediction = 'Red Oak'
    elif predicted_class == 4:
        prediction = 'Silver Birch'
    elif predicted_class == 5:
        prediction = "Sweet Gum"

    print(img + " = " + prediction)
```

```python
test_image = image.load_img(r'./LeavesIdentifier/Unkown/Leaf/Leaf_12.jpg', target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis = 0)
predict = Model.predict(test_image)
if predict[0][0] == 1:
    prediction = 'Cherry Blossom'
elif predict[0][1] == 1:
    prediction = 'Gingko'
elif predict[0][2] == 1:
    prediction = 'Japanese Maple'
elif predict[0][3] == 1:
    prediction = 'Red Oak'
elif predict[0][4] == 1:
    prediction = 'Silver Birch'
elif predict[0][5] == 1:
    prediction = "Sweet Gum"
print(prediction)
```

## Model Accuracy

```python
print(history.history.keys())
import matplotlib.pyplot as plt

# Plotting learning curves
plt.figure(figsize=(12, 6))

# Plot training & validation accuracy values
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'])  
plt.plot(history.history['val_accuracy']) 
plt.title('Model accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation-Test'], loc='upper left')

# Plot training & validation loss values
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation-Test'], loc='upper left')

plt.tight_layout()
plt.show()
```

```python
test_accuracy = Model.evaluate(test_set)
print(f'Test Accuracy: {test_accuracy[1]}')
```

**The CNN classification model was found to be 91.05 % accurate on test set.**
